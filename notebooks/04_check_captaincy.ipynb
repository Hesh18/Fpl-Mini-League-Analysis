```python
import duckdb

con = duckdb.connect("../data/warehouse.duckdb", read_only=True)
```

```python
con.sql("""
    SELECT
        team_name,
        count(*) FILTER (WHERE changed_captain) AS captain_changes,
        count(*) FILTER (WHERE captain_points > crowd_captain_points) AS beat_crowd,
        count(*) FILTER (WHERE captain_points < crowd_captain_points) AS behind_crowd
    FROM mart_captain_decisions
    GROUP BY team_name
    ORDER BY captain_changes DESC, team_name
""").df()
```
