```python
import duckdb

con = duckdb.connect("../data/warehouse.duckdb", read_only=True)
```

```python
con.sql("""
    SELECT player_name, position, good_games, full_games, best_game, is_ranked
    FROM mart_player_consistency
    WHERE player_name IN ('Muharemović', 'B.Fernandes', 'Haaland')
""").df()
```

```python
print(con.sql("""
    SELECT position, player_name, good_games || ' of ' || full_games AS record, best_game
    FROM mart_player_consistency
    WHERE is_ranked
    QUALIFY row_number() OVER (PARTITION BY position ORDER BY good_game_share DESC, full_games DESC, player_name) = 1
    ORDER BY position
""").df())
con.close()
```
