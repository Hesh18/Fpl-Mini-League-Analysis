```python
import duckdb

con = duckdb.connect("../data/warehouse.duckdb", read_only=True)
```

```python
con.sql("""
    SELECT gameweek, count(*) AS managers, median(points - transfer_cost) AS median_score
    FROM stg_manager_gameweeks
    WHERE gameweek <= 10
    GROUP BY gameweek
    ORDER BY gameweek
""").df()
```

```python
print(con.sql("""
    SELECT team_name, top_half_weeks, gameweeks_played, longest_run, current_run
    FROM mart_top_half_weeks
    ORDER BY top_half_share DESC, longest_run DESC, team_name
""").df())
con.close()
```
